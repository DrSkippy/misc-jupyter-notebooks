```python
import scipy.special
import numpy as np
from matplotlib import pyplot as plt

def gDist(x, alpha, beta):
    return (beta**(-alpha))*(x**(alpha-1.))*np.exp(-x/beta) / scipy.special.gamma(alpha)
    
def dExp(x, alpha, beta, A=1):
    return A*(1. - np.exp(-alpha*x))*np.exp(-beta*x)
```

```python
t = np.linspace(0,8,100)
alpha=0.1
beta=0.8
plt.plot(t, dExp(t, alpha, beta, A=9.2))
alpha=2.4
beta=0.8
plt.plot(t, gDist(t, alpha, beta))
```
