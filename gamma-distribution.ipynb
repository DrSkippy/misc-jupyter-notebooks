# Gamma Distribution vs a Double Exponential

Two right-skewed shapes that rise from zero, peak, and decay:

- the gamma density $f(x) = \dfrac{x^{\alpha-1} e^{-x/\beta}}{\Gamma(\alpha)\,\beta^{\alpha}}$ (shape $\alpha$, scale $\beta$), from `scipy.stats.gamma`
- a "double exponential" rise-and-decay curve $g(x) = A\,(1 - e^{-\alpha x})\,e^{-\beta x}$ (rate constants $\alpha$, $\beta$)

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats


def double_exp(x, alpha, beta, A=1.0):
    """Rise with rate alpha, decay with rate beta, amplitude A."""
    return A * (1 - np.exp(-alpha * x)) * np.exp(-beta * x)


t = np.linspace(0, 8, 200)
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(t, double_exp(t, alpha=0.1, beta=0.8, A=9.2), label=r"double exponential ($\alpha$=0.1, $\beta$=0.8, A=9.2)")
ax.plot(t, stats.gamma.pdf(t, a=2.4, scale=0.8), label=r"gamma pdf ($\alpha$=2.4, $\beta$=0.8)")
ax.set_xlabel("x")
ax.legend()
plt.show()
```
